```python
import pandas as pd

df = pd.DataFrame({
    'job_id': [1, 2, 3, 4],
    'job_title': ['Data Analyst', 'Data Scientist', 'Data Engineer', 'Data Analyst'],
    'salary': [65000, 95000, 88000, 70000],
    'skills': ['Python, SQL', 'Python, R, SQL', 'SQL, Spark', 'Excel, SQL']
})
```

```python
df.to_csv('jobs.csv', index = False)
```

```python
df.to_excel('jobs.xlsx')
```

```python
df['salary_k'] = df['salary'].apply(lambda x: (x/1000))
```

```python
df
```

```python
def categorize_salary(salary):
    if salary < 70000:
        print('low')
    else:
        print('high')
```

```python
df['salary_level'] = df['salary'].apply(categorize_salary)
```

```python
df['skills'] = df['skills'].str.split(', ')
df_exploded = df.explode('skills')
```

```python
df_exploded
```

```python
skills_count = df_exploded.skills.value_counts()
```

```python
import matplotlib.pyplot as plt

df.plot(kind = 'bar', x='job_title', y='salary', figsize = (10,5))
plt.xticks(rotation = 45, ha = 'right')
plt.tight_layout()
```

```python
skills_count.plot(kind = 'pie', autopct = '%1.1f%%')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    'job_title': ['Data Analyst', 'Data Scientist', 'Data Engineer', 'Data Analyst',
                  'Data Scientist', 'Data Engineer', 'Data Analyst', 'Data Scientist'],
    'salary': [65000, 95000, 88000, 70000, 105000, 92000, 68000, 99000]
})
```

```python
df
```

```python
df['salary'].plot(kind = 'hist', bins = 5)
```

```python
df['salary'].plot(kind = 'box')
```

```python
df.boxplot(column='salary', by='job_title')
```

```python
avg_salary = df.groupby('job_title')['salary'].mean()
avg_salary.plot(kind='bar', color='green', figsize=(8,5))
plt.grid()
plt.show()
```

```python
ax = avg_salary.plot(kind='bar')
for i, v in enumerate(avg_salary):
    ax.text(i, v, str(round(v)), ha='center', va='bottom')
plt.show()
```
