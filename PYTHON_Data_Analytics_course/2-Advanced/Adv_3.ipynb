```python
# Importing libraries

import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Loading data

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data clean up

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
# cuz this date column was type str, now changed to type datetime
```

```python
df_us = df[df['job_country'] == 'United States'].copy()
df_us
```

```python
df_us['job_posted_month'] = df_us['job_posted_date'].dt.strftime('%B')
# creates a new column, based on the changing of existing column
```

```python
df_us
```

```python
df_us_pivot = df_us.pivot_table(index = 'job_posted_month', columns = 'job_title_short', aggfunc = 'size')
```

```python
df_us_pivot
```

```python
df_us_pivot.reset_index(inplace = True)
```

```python
df_us_pivot
```

```python
df_us_pivot['month'] =pd.to_datetime(df_us_pivot['job_posted_month'], format = '%B').dt.month
```

```python
df_us_pivot
```

```python
df_us_pivot.sort_values('month', inplace = True)
df_us_pivot.set_index('job_posted_month', inplace = True)
df_us_pivot.drop(columns = 'month', inplace = True)
```

```python
df_us_pivot
```

```python
df_us_pivot.plot(kind = 'line')
```

```python
df_da = df[df['job_title_short'] == 'Data Analyst'].copy()
```

```python
df_da['job_month'] = df_da['job_posted_date'].dt.month
```

```python
df_da_exploded = df_da.explode('job_skills')
```

```python
df_da_pivot = df_da_exploded.pivot_table(index = 'job_month', columns='job_skills',  aggfunc = 'size', fill_value = 0)

df_da_pivot
```

```python
# Create a total row
df_da_pivot.loc['Total'] = df_da_pivot.sum()

# Order columns by total count
df_da_pivot = df_da_pivot[df_da_pivot.loc['Total'].sort_values(ascending =  False).index]

# Drop total row
df_da_pivot = df_da_pivot.drop('Total')

df_da_pivot
```

```python
df_da_pivot = df_da_pivot.reset_index()

df_da_pivot['job_posted_month'] = df_da_pivot['job_month'].apply(lambda x: pd.to_datetime(x, format='%m').strftime('%b'))
# new column job_posted_month where Jan Feb Mar ...

df_da_pivot = df_da_pivot.set_index('job_posted_month')
df_da_pivot = df_da_pivot.drop(columns = 'job_month')

df_da_pivot
```

```python
df_da_pivot.iloc[:, :5].plot(kind='line')

plt.title('Top 5 Skills for Data Analysts per Month')
plt.ylabel('Count')
plt.xlabel('')
plt.show()
```
