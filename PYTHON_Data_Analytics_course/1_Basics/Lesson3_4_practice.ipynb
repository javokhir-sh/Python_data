```python
# 7. Create a small DataFrame from a dictionary with columns name, age, city (3-4 rows).
```

```python
import pandas as pd
```

```python
dicts = [
    {'name': 'John', 'age': 25, 'city': 'New York'},
    {'name': 'Alice', 'age': None, 'city': 'Los Angeles'},
    {'name': 'Bob', 'age': 22, 'city': None},
    {'name': 'Luke', 'age': 31, 'city': 'Urgench'},
    {'name': 'Anna', 'age': 28, 'city': 'London'},
    {'name': 'Tom', 'age': None, 'city': 'Paris'},
    {'name': 'Kate', 'age': 29, 'city': 'Berlin'}
    ]
```

```python
df = pd.DataFrame(dicts)
```

```python
# 10. Given a DataFrame with some missing values (NaN), check how many missing values exist per column.
```

```python
df['age'].notna().value_counts()
```

```python
df['city'].notna().value_counts()
```

```python
df['name'].notna().value_counts()
```

```python
df.isna().sum()    # counts none values for each column
```

```python
# 11. Drop rows with any missing values. Then, separately, fill missing values with 0 instead.
```

```python
df_d = df.dropna().copy()
df_d['age'] = df_d['age'].astype(int)
```

```python
df_d
```

```python
df_f = df.fillna(0).copy()     # gives 0 for none values
df_f['age'] = df_f['age'].astype(int)
```

```python
df_f
```
