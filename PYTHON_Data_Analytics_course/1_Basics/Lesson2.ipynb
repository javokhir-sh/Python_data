```python
class JohnSalary:
  def __init__(self, base_salary, bonus_rate=.1, symbol = '$'):
    self.base_salary = base_salary
    self.bonus_rate = bonus_rate
    self.symbol = symbol
    self.total_salary = base_salary * (bonus_rate + 1)
    self.bonus = (self.total_salary - base_salary) / base_salary

  def __repr__(self):
    return f'{self.symbol}{self.base_salary:,.1f}'

  def calculate_salary(self):
    return f'{self.symbol}{self.total_salary:,.1f}'

  def calculate_bonus(self):
    return f'{self.symbol}{self.bonus:,.1f}'
```
