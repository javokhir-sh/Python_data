```python
# Labeling ##
```

```python
# 4. Take your line chart from #1 and add a title, x-axis label, and y-axis label.

x = [1,2,3,4,5]
y = [1,4,9,16,25]
plt.plot(x,y)
plt.title('Numbers')
plt.xlabel('Xaaa')
plt.ylabel('Yaaa')
```

```python
# 5. Plot two lines on the same chart (e.g. y1 = [1,4,9,16,25] and y2 = [1,2,3,4,5]) and add a legend distinguishing them.

y1 = [1,4,9,16,25]
y2 = [1,2,3,4,5]

plt.plot(y1, label = 'y1')
plt.plot(y2, label = 'y2')
plt.legend()
```

```python
# 6. Change the color and line style of one of your plots (e.g. dashed red line).
plt.plot(x, y, color = 'red', linestyle = '--')

```

```python
plt.plot(y1, label = y1, color = 'blue', linestyle = ':')
plt.plot(y2, label = y2, color = 'green', linestyle = '-.')
plt.legend()
```

```python
# 7. Using a small DataFrame with a numeric column, call .plot() directly on it (pandas has built-in plotting via matplotlib).
```

```python
import pandas as pd
```

```python
dicts = [
    {'name': 'John', 'age': 25, 'city': 'New York'},
    {'name': 'Alice', 'age': None, 'city': 'Los Angeles'},
    {'name': 'Bob', 'age': 22, 'city': None},
    {'name': 'Luke', 'age': 31, 'city': 'Urgench'},
    {'name': 'Anna', 'age': 28, 'city': 'London'},
    {'name': 'Tom', 'age': None, 'city': 'Paris'},
    {'name': 'Kate', 'age': 29, 'city': 'Berlin'}
    ]
```

```python
df = pd.DataFrame(dicts)
```

```python
df['age'].plot()
```

```python
# 8. Group df by job_title and find the average salary per job title. Plot the result as a bar chart, with a title and a y-axis label ("Salary").
```

```python
data = {
    'job_title': ['Data Analyst', 'Data Scientist', 'Data Analyst', 'ML Engineer', 'Data Scientist', 'Data Analyst'],
    'salary': [65000, 95000, 70000, 110000, 98000, 68000]
}
df = pd.DataFrame(data)
```

```python
df_gr = df.groupby('job_title')['salary'].mean()
df_gr.plot(kind = 'bar')
plt.title('Average Salary by Job Title')
plt.ylabel('Salary')
```

```python
# 9. Plot a histogram of the salary column to see how salaries are distributed across all rows. Add a title and an x-axis label ("Salary").
```

```python
df['salary'].plot(kind = 'hist')
plt.title('Salary Distribution')
plt.xlabel('Salary')
```

```python
# 10. Take the groupby result from earlier practice (average age per city) and plot it directly as a bar chart.
```

```python
df_grouped = df.groupby('city')['age'].mean()
df_grouped.plot(kind = 'bar')
```
