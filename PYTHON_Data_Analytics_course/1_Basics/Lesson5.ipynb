```python
job_counts = df.job_title_short.value_counts()
job_counts = job_counts.sort_values()
plt.barh(job_counts.index, job_counts)
```

```python
job_counts = df.job_title_short.value_counts(ascending = True)
job_counts = job_counts.sort_values()
plt.barh(job_counts.index, job_counts)
plt.show()   #  optional
```

```python
plt.bar(job_counts.index, job_counts)
plt.title('Johns data')
plt.ylabel('Counted values')
plt.xlabel('Job titles')
plt.xticks(rotation=45, ha = 'right')
plt.show()
```

```python
job_salaries = df.groupby('job_title_short')['salary_year_avg'].median().sort_values()
```

```python
job_salaries.plot(kind='barh')
plt.title('Johns data')
plt.ylabel('Job roles')
plt.xlabel('Salaries in USD($)')
plt.show()
```
